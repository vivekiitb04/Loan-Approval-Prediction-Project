##Label Encoding the categorical variables

```python

print(df['education'].unique())
print(df['self_employed'].unique())
print(df['loan_status'].unique())
```

```python
df['education'] = df['education'].replace({
    ' Not Graduate': 0,
    ' Graduate': 1
})

df['self_employed'] = df['self_employed'].replace({
    ' No': 0,
    ' Yes': 1
})

df['loan_status'] = df['loan_status'].replace({
    ' Rejected': 0,
    ' Approved': 1
})
```

#Train Test Split

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(df.drop('loan_status',axis=1),df['loan_status'],test_size=0.2,random_state=2)
```

```python
X_train.shape
```

```python
y_train.shape
```

```python
X_test.shape
```

#Model Building



1.   Decision Tree Classifier
2.   Random Forest Classifier
3.   Logistic Regression




###**Decision Tree Classifier**

```python
from sklearn.metrics import accuracy_score
from sklearn.tree import DecisionTreeClassifier

dtr=DecisionTreeClassifier()
```

```python
dtr.fit(X_train,y_train)
```

```python
#Training Score
dtr.score(X_train,y_train)
```

```python
#Predicting the Loan Approval Status
dtr_pred=dtr.predict(X_test)
```

```python
print("Accuracy score = ", accuracy_score(y_test,dtr_pred))
```

##**Random Forest Classifier**

```python
from sklearn.ensemble import RandomForestClassifier
rfc=RandomForestClassifier()
```

```python
rfc.fit(X_train,y_train)
```

```python
#Training Score
print("Training Score = ", rfc.score(X_train,y_train))
```

```python
#Prediction

rfc_pred=rfc.predict(X_test)
print("Accuracy Score = ",accuracy_score(y_test,rfc_pred))
```

##Logistic Regression

```python
from sklearn.linear_model import LogisticRegression
lor=LogisticRegression()
```

```python
lor.fit(X_train,y_train)
```

```python
#Training Score
print("Training score =",lor.score(X_train,y_train))
```

```python
#Prediction
lor_pred=lor.predict(X_test)

print("Accuracy Acore=",accuracy_score(y_test,lor_pred))
```

#Model Metrics

###Decision Tree Classifier

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

print("Accuracy :", accuracy_score(y_test, dtr_pred))
print("Precision:", precision_score(y_test, dtr_pred))
print("Recall   :", recall_score(y_test, dtr_pred))
print("F1 Score :", f1_score(y_test, dtr_pred))
```

##Random Forest Classifier

```python

print("Accuracy :", accuracy_score(y_test, rfc_pred))
print("Precision:", precision_score(y_test, rfc_pred))
print("Recall   :", recall_score(y_test, rfc_pred))
print("F1 Score :", f1_score(y_test, rfc_pred))
```

##Logistic Regression

```python
print("Accuracy :", accuracy_score(y_test, lor_pred))
print("Precision:", precision_score(y_test, lor_pred))
print("Recall   :", recall_score(y_test, lor_pred))
print("F1 Score :", f1_score(y_test, lor_pred))
```

#Model Evaluation
